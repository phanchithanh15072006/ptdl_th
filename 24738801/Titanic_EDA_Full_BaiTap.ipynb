# PHẦN B — DESCRIPTIVE STATISTICS
## Group 1 — Central Tendency

````python
print("Do luong xu huong tap trung (Central tendency)")
print("mean (trung binh)")
print("Tuổi trung bình:", df['age'].mean())
print("Giá vé trung bình :", df['fare'].mean())

print("median (trung vi)")
print("Tuổi trung vị :", df['age'].median())
print("Giá vé trung vị :", df['fare'].median())

print("mode (xuat hien nhieu nhat)")
print("Độ tuổi phổ biến nhất:", df['age'].mode()[0])
print("Cảng khởi hành  đông nhất:", df['embarked'].mode()[0])
print("Hạng vé  được mua nhiều nhất:", df['pclass'].mode()[0])
````

## Group 2 — Dispersion

````python
print("Do luong phann tan (dispersion)")
cols = ['age', 'fare']


print("range(khoang bien thien max- min)")
for col in cols:
    rng = df[col].max() - df[col].min()
    print(f"Khoảng biến thiên của '{col}': {rng:.2f}")


print("variance(phuong sai)")
print(df[cols].var())


print("standard deviation (do lech chuan)")
print(df[cols].std())


````

## Group 3 — Location and Shape

````python

cols = ['age', 'fare']


print("quantiles (cac diem phan vi)")

print(df[cols].quantile([0.25, 0.50, 0.75, 0.90]))


print("skewness( do lech)")
print(df[cols].skew())


print("kurtosis (do nhon)")
print(df[cols].kurtosis())
````

# PHẦN C — DEFINE THE QUESTION

## Câu hỏi 1: Hạng vé nào có tỷ lệ sống sót cao nhất, chênh lệch bao nhiêu so với hạng thấp nhất?

````python

survived = df.groupby(by='pclass')['survived'].mean()

print("hang ve co ti le song cao nhat:",survived.idxmax())
print("chenh lech so voi hang thap nhat la:",(survived.max()-survived.min())*100,"%")
````

## Câu hỏi 2: Giới tính hay hạng vé ảnh hưởng đến sống sót mạnh hơn?

````python

# 1. Tỷ lệ theo Giới tính
surv_sex = df.groupby('sex')['survived'].mean() * 100
diff_sex = surv_sex.max() - surv_sex.min()

# 2. Tỷ lệ theo Hạng vé (lấy lại kết quả từ câu 1)
surv_class = df.groupby('pclass')['survived'].mean() * 100
diff_class = surv_class.max() - surv_class.min()

# 3. Kết hợp cả hai yếu tố bằng bảng Pivot
pivot_combined = df.pivot_table(
    index='sex', 
    columns='pclass', 
    values='survived', 
    aggfunc='mean'
) * 100

print(f"Chênh lệch sống sót theo Giới tính: {diff_sex:.2f}%")
print(f"Chênh lệch sống sót theo Hạng vé: {diff_class:.2f}%\n")
print(" Bảng tỷ lệ sống sót chi tiết (%) ")
print(pivot_combined.round(2))
````

## Câu hỏi 3: Vé đắt hơn có thực sự sống sót cao hơn không?

````python
price = df.groupby(by="survived")['fare'].mean()

print(price)
````

## Câu hỏi 4: Gia đình đông người có ảnh hưởng đến khả năng sống sót không?

````python

df['family_size'] = df['sibsp'] + df['parch'] + 1


family = df.groupby(by=["family_size"])["survived"].mean()


print(family )
````

## Câu hỏi 5: Cảng lên tàu (embark_town) nào có tỷ lệ sống sót cao nhất?

````python

embark_towns = df.groupby(by=["embark_town"])["survived"].mean()
print(embark_towns)
````
